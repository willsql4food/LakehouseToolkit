```python
cat = spark.getActiveSession().catalog

for c in cat.listCatalogs():
    print(f"Catalog: {c}")

for d in cat.listDatabases():
    print(f"Database: {d.name}")
    spark.sql(f"use schema {d.name}")
    for t in cat.listTables():
        print(f"\tTable: {t.namespace[0]}.{t.name}")
```

```python
for t in cat.listTables():
    if t.name.startswith('tbl_hierarchy'):
        print(f"spark.sql('drop table {t.namespace[0]}.{t.name}')")
```
