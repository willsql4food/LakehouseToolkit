```python
tbls = spark.sql("show tables").collect()

for t in tbls:
    if t.tableName.startswith('ga_'):
        i = 0
        for c in spark.sql(f"describe table {t.tableName}").collect():
            if c.data_type == "int":
                if i == 0:
                    print(f"spark.sql(\"ALTER TABLE {t.tableName} SET TBLPROPERTIES ('delta.enableTypeWidening' = 'true')\")")
                    i += 1
                print(f"spark.sql(\"alter table {t.tableName} alter column {c.col_name} type bigint\")")

```
